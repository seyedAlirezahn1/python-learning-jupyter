oop(inheritance)

```python
class Person:
  salm = 1000000
  def __init__(self, name="noname", age=0, gender=False):
      self.name = name
      self.age = age
      self.gender = gender  

class Student(Person):
   def __init__(self, name="noname", age=0, gender=False, gpa= 0):
      super().__init__(name, age, gender)
      self.gpa = gpa

p = Student("sadra",23,True,20)
print(p.gpa)
print(p.salm)
```

```python
class Person:
  __salm = 1000000
  def __init__(self, name="noname", age=0, gender=False):
      self.name = name
      self.age = age
      self.gender = gender  

class Student(Person):
   def __init__(self, name="noname", age=0, gender=False, gpa= 0):
      super().__init__(name, age, gender)
      self.gpa = gpa

p = Student("sadra",23,True,20)
print(p.gpa)
#print(p.salm) ***incorrect***
```

```python
import matplotlib.pyplot as plt
import numpy as np

def f(x,y):
  return x**2+y**2

x = np.linspace(-10, 10, 400)
y = np.linspace(-10, 10, 400)
X, Y = np.meshgrid(x,y)

Z = f(X,Y)

ax = plt.axes(projection="3d")
ax.contour3D(X, Y, Z, 400, cmap="copper")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("z")
plt.show()
```
