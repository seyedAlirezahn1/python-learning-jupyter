In the third test case, the left part may be [1]

```python
a = int(input())

for _ in range(a):
    b = int(input())
    arr = list(map(int, input().split()))
    
    one = [0]
    two = [0]
    three = [0]
    
    for num in arr:
        one.append(one[-1])
        two.append(two[-1])
        three.append(three[-1])
        
        if num == 1:
            one[-1] += 1
        elif num == 2:
            two[-1] += 1
        else:
            three[-1] += 1
    
    found = False
    

    for i in range(b - 2):
        ones_left = one[i + 1]
        others_left = two[i + 1] + three[i + 1]
        
        if ones_left < others_left:
            continue

        for j in range(i + 1, b - 1):
            ones_mid = one[j + 1] - one[i + 1]
            twos_mid = two[j + 1] - two[i + 1]
            threes_mid = three[j + 1] - three[i + 1]
            
            if ones_mid + twos_mid >= threes_mid:
                found = True
                break
        
        if found:
            break
    
    print("YES" if found else "NO")
```
